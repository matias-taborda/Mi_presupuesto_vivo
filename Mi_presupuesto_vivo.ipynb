```python
#### from datetime import datetime

# =========================
# DATOS DEL USUARIO
# =========================
usuario = input("Ingrese nombre del usuario: ")
fecha = datetime.now().strftime("%d/%m/%Y %H:%M")

print("\n📅 Registro financiero")
print("Usuario:", usuario)
print("Fecha:", fecha)

# =========================
# INGRESOS
# =========================
print("\n📈 INGRESO DE INGRESOS")

ingresos = {
    "Haberes": float(input("Haberes: $")),
    "Extraordinarios": float(input("Extraordinarios: $")),
    "Intereses": float(input("Intereses: $")),
    "Ayuda económica": float(input("Ayuda económica: $"))
}

total_ingresos = sum(ingresos.values())

# =========================
# AHORRO METAS
# =========================
porcentaje_metas = float(input("\nIngrese porcentaje de ahorro metas (%): ")) / 100
ahorro_metas = ingresos["Haberes"] * porcentaje_metas

# =========================
# COLCHÓN FINANCIERO
# =========================
porcentaje_colchon = float(input("\nIngrese porcentaje de colchón financiero (%): ")) / 100
colchon_financiero = ingresos["Haberes"] * porcentaje_colchon

# =========================
# GASTOS
# =========================
print("\n📉 INGRESO DE GASTOS")

gastos = {
    "Alquiler": float(input("Alquiler: $")),
    "Servicios": float(input("Servicios: $")),
    "Gastos personales": float(input("Gastos personales: $")),
    "Deudas": float(input("Deudas: $"))
}

total_gastos = sum(gastos.values())
gastos_totales = total_gastos + ahorro_metas + colchon_financiero 

# =========================
# BALANCE
# =========================
print("\n📊 BALANCE FINAL")

balance = total_ingresos - total_gastos - ahorro_metas - colchon_financiero

print (f"BALANCE FINAL: ${balance:.2f}")

# =========================
# PORCENTAJES MANUALES
# =========================
print("\n📊 INGRESO DE PORCENTAJES DE GASTOS")

porcentajes = {}
for categoria in gastos:
    porcentaje = float(input(f"Porcentaje asignado a {categoria} (%): ")) / 100
    porcentajes[categoria] = porcentaje

porcentajes["Ahorro metas"] = porcentaje_metas

porcentajes["Colchón financiero"] = porcentaje_colchon

# Límites recomendados
limites = {
    "Alquiler": 0.42,
    "Servicios": 0.15,
    "Gastos personales": 0.10,
    "Deudas": 0.00,
    "Ahorro metas": 0.15,
    "Colchón financiero": 0.10
}

minimos = ["Ahorro metas", "Colchón financiero"]

# =========================
# RESUMEN
# =========================
print("\n==============================")
print("📊 RESUMEN FINANCIERO")
print("==============================")

print(f"\n👤 Usuario: {usuario}")
print(f"📅 Fecha: {fecha}")

print("\n📈 INGRESOS")
for k, v in ingresos.items():
    print(f"{k}: $ {v:.2f}")
print(f"TOTAL INGRESOS: $ {total_ingresos:.2f}")

print("\n📉 GASTOS")
for k, v in gastos.items():
    print(f"{k}: $ {v:.2f}")

print(f"Ahorro metas: $ {ahorro_metas:.2f}")
print(f"Colchón financiero: $ {colchon_financiero:.2f}")
print(f"TOTAL GASTOS: $ {gastos_totales:.2f}")

# =========================
# EVALUACIÓN POR PORCENTAJE
# =========================
print("\n📊 EVALUACIÓN DE GASTOS POR PORCENTAJE\n")

for categoria, porcentaje in porcentajes.items():
    pct = porcentaje * 100
    limite = limites.get(categoria, None)

    if categoria in minimos:
        # Ahorros: se evalúa por mínimo
        if porcentaje < limite:
            falta = (limite - porcentaje) * 100
            print(f"⚠️ {categoria}: {pct:.1f}% (faltan {falta:.1f}% para el mínimo recomendado)")
        else:
            print(f"✅ {categoria}: {pct:.1f}% (mínimo cubierto)")
    else:
        # Gastos comunes: se evalúa por máximo
        if limite is not None and porcentaje > limite:
            exceso = (porcentaje - limite) * 100
            print(f"⚠️ {categoria}: {pct:.1f}% (exceso {exceso:.1f}%)")
        else:
            print(f"✅ {categoria}: {pct:.1f}%")

# =========================
# ALERTA FINAL
# =========================
if gastos_totales > total_ingresos:
    print("\n⚠️ ALERTA: Gastos mayores a los ingresos")
else:
    print("\n✅ OK: Finanzas dentro de los ingresos")
```
